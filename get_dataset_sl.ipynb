```python
import copernicusmarine
import os 
import xarray as xr
from copernicusmarine import subset
# !copernicusmarine login
```

```python
nome_product = "cmems_mod_glo_phy_anfc_0.083deg_PT1H-m"
path = "/home/samuele/Desktop/aips/data/main_data/copernicus"

nc_output = os.path.join(path, nome_product+".nc")
csv_output = os.path.join(path, nome_product+".csv")
```

```python
subset(
    dataset_id = nome_product,  # ← lo sostituirai con il vero ID (vedi sotto)
    minimum_longitude=13.30,
    maximum_longitude=13.85,
    minimum_latitude=45.55,
    maximum_latitude=45.85,
    start_datetime="2024-06-01T00:00:00",
    end_datetime="2024-06-10T00:00:00",
    #variables=["thetao"],  # esempio: temperatura del mare
    output_filename=nc_output
)
```

```python
ds = xr.open_dataset(nc_output)
```

```python
df = ds.to_dataframe().reset_index()
```

```python
df.to_csv(csv_output, index=False)
```
