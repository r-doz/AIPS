```python
# read data "vessel_per_day.csv" and plot a histogram of the number of vessels per day
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import os
import sys
import matplotlib.dates as mdates


# read data
data = pd.read_csv("vessel_per_day.csv", sep=";", parse_dates=["date"])

# sort data by date
data = data.sort_values(by="date")

# set the figure size
plt.figure(figsize=(10, 6))

# plot the histogram
sns.barplot(x="date", y="all_vessel", data=data, color="blue")

# format the x-axis to show dates properly
#plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
plt.gca().xaxis.set_major_locator(mdates.DayLocator(interval=10))
plt.xticks(rotation=45)

# set labels and title
plt.xlabel("Date")
plt.ylabel("Number of Vessels")
plt.title("Number of Vessels per Day")

# adjust layout to prevent clipping of tick-labels
plt.tight_layout()

# show the plot
plt.show()
```
